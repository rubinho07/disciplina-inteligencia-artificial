Exercicio 21 - A busca binaria procura um elemento em uma lista ordenada, com isso dividindo a área de busca pela metade a cada repetição .

```python
def busca_binaria(lista, elemento):
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2

        if lista[meio] == elemento:
            return meio
        elif lista[meio] < elemento:
            inicio = meio + 1
        else:
            fim = meio - 1

    return -1


numeros = [2, 5, 8, 12, 16, 23, 38, 45]
procurado = int(input("Digite o número que deseja procurar: "))

posicao = busca_binaria(numeros, procurado)

if posicao != -1:
    print(f"O número foi encontrado na posição {posicao}.")
else:
    print("O número não foi encontrado.")
```

Exercicio 22 - o cifra ele substitui cada letra por outra que esta em posições a frente no alfabeto.

```python
def criptografar(mensagem, deslocamento):
    mensagem_cifrada = ""

    for caractere in mensagem:
        if caractere.isalpha():
            if caractere.isupper():
                codigo_inicial = ord("A")
            else:
                codigo_inicial = ord("a")

            posicao = ord(caractere) - codigo_inicial
            nova_posicao = (posicao + deslocamento) % 26
            nova_letra = chr(codigo_inicial + nova_posicao)

            mensagem_cifrada += nova_letra
        else:
            mensagem_cifrada += caractere

    return mensagem_cifrada


mensagem = input("Digite a mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

resultado = criptografar(mensagem, deslocamento)

print("Mensagem criptografada:", resultado)
```

Exercicio 24 - Aqui usei a função true , quando os digitos calculados são iguais aos dois ultimos digitos do cpf ele retorna.

```python
def validar_cpf(cpf):
    cpf = "".join(caractere for caractere in cpf if caractere.isdigit())

    if len(cpf) != 11:
        return False

    if cpf == cpf[0] * 11:
        return False

    soma = 0

    for indice in range(9):
        soma += int(cpf[indice]) * (10 - indice)

    resto = soma % 11

    if resto < 2:
        primeiro_digito = 0
    else:
        primeiro_digito = 11 - resto

    soma = 0

    for indice in range(10):
        soma += int(cpf[indice]) * (11 - indice)

    resto = soma % 11

    if resto < 2:
        segundo_digito = 0
    else:
        segundo_digito = 11 - resto

    return (
        primeiro_digito == int(cpf[9])
        and segundo_digito == int(cpf[10])
    )


cpf = input("Digite um CPF: ")

if validar_cpf(cpf):
    print("CPF válido.")
else:
    print("CPF inválido.")
```

```python
import random

palavras = ["python", "computador", "programacao", "algoritmo"]
palavra_secreta = random.choice(palavras)

letras_corretas = set()
letras_erradas = set()
tentativas = 6

print("Bem-vindo ao jogo da forca!")

while tentativas > 0:
    palavra_exibida = ""

    for letra in palavra_secreta:
        if letra in letras_corretas:
            palavra_exibida += letra + " "
        else:
            palavra_exibida += "_ "

    print("\nPalavra:", palavra_exibida)
    print("Letras erradas:", letras_erradas)
    print("Tentativas restantes:", tentativas)

    if all(letra in letras_corretas for letra in palavra_secreta):
        print("Parabéns! Você descobriu a palavra.")
        break

    palpite = input("Digite uma letra: ").lower()

    if len(palpite) != 1 or not palpite.isalpha():
        print("Digite somente uma letra.")
        continue

    if palpite in letras_corretas or palpite in letras_erradas:
        print("Essa letra já foi utilizada.")
        continue

    if palpite in palavra_secreta:
        letras_corretas.add(palpite)
        print("Você acertou uma letra!")
    else:
        letras_erradas.add(palpite)
        tentativas -= 1
        print("Essa letra não está na palavra.")

else:
    print("Suas tentativas acabaram.")
    print("A palavra era:", palavra_secreta)
```
